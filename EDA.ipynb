```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("D:\yt series\Diabetes_Dataset.csv")
df.head(50)
df.tail(50)
```

```python
df.shape
df.info()
df.describe()
```

```python
df['Gender'] = df['Gender'].str.capitalize()
print(df)
```

```python
df.isnull().sum()
```

```python
df=df.fillna(df.mean(numeric_only=True))
print(df)
```

```python
print(df.duplicated().sum())
df.drop_duplicates(inplace=True)
```

```python
df.hist(figsize=(12,10))
plt.show()
```

```python
plt.plot(df['Age'], df['BloodPressure'])
plt.xlabel("Age")
plt.ylabel("BloodPressure")
plt.title("Age vs BloodPressure")
plt.show()
plt.plot(df['BMI'], df['Age'])
plt.xlabel("BMI")
plt.ylabel("Age")
plt.title("BMI vs Age")
plt.show()
```

```python
sns.countplot(x='Gender', data=df)
plt.show()
sns.countplot(x='PhysicalActivity', data=df)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df[['BMI','GlucoseLevel','BloodPressure']])
plt.show()
```

```python
Q1 = df['BMI'].quantile(0.25)
Q3 = df['BMI'].quantile(0.75)
IQR = Q3 - Q1
df = df[(df['BMI'] >= Q1 - 1.5*IQR) & (df['BMI'] <= Q3 + 1.5*IQR)]
df
```

```python
plt.scatter(df.index, df['GlucoseLevel'])
plt.title("Outliers in Glucose Level")
plt.show()
```

```python
sns.boxplot(x='Gender', y='BMI', data=df)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.show()
```

```python
df_encoded = pd.get_dummies(df, drop_first=True)
df_encoded.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
num_cols = ['Age','BMI','BloodPressure','GlucoseLevel','Insulin']
df_encoded[num_cols] = scaler.fit_transform(df_encoded[num_cols])
df_encoded[num_cols]
```

```python
from scipy import stats
z_scores = np.abs(stats.zscore(df['BMI']))
df_clean = df[z_scores < 3]
print("Original Shape:", df.shape)
print("After Removing Outliers:", df_clean.shape)
```
